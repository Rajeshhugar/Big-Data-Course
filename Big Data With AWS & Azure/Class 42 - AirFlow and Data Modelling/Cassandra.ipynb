## Chapter 1: Cassandra Fundamentals
### 1.1 What is Cassandra?

**Definition:**  
Apache Cassandra is a distributed NoSQL database designed to handle massive amounts of data across multiple nodes with no single point of failure.  

**Key Features:**  
```mermaid
graph TD
    A[Cassandra] --> B[Linear Scalability]
    A --> C[High Availability]
    A --> D[Fault Tolerance]
    A --> E[Tunable Consistency]
    A --> F[Multi-DC Support]
```

**Use Cases:**  
- Time-series data (IoT, metrics)  
- Product catalogs  
- Messaging systems  
- Recommendation engines

### 1.2 Architecture Deep Dive
**Peer-to-Peer Ring Topology:**  
```mermaid
graph LR
    N1[Node 1] --> N2[Node 2]
    N2 --> N3[Node 3]
    N3 --> N4[Node 4]
    N4 --> N1
```

**Key Components:**  
1. **Partitioner:** Maps data to nodes using consistent hashing  
2. **Replication Factor:** Number of data copies (default=3)  
3. **Gossip Protocol:** Node communication mechanism  
4. **Hinted Handoff:** Temporary storage for unavailable nodes

## Chapter 2: Data Modeling
### 2.1 Cassandra vs Relational Databases
| Aspect | Cassandra | RDBMS |  
|--------|-----------|-------|  
| Schema | Flexible | Rigid |  
| Joins | Not supported | Supported |  
| Scaling | Horizontal | Vertical |

### 2.2 Core Concepts
**1. Keyspace:** Database equivalent  
```cql
CREATE KEYSPACE ecommerce 
WITH replication = {'class': 'NetworkTopologyStrategy', 'DC1': 3};
```

**2. Table Structure:**  
```cql
CREATE TABLE products (
    category TEXT,
    product_id UUID,
    name TEXT,
    price DECIMAL,
    PRIMARY KEY ((category), product_id)
);
```

**3. Primary Key Components:**  
- **Partition Key:** Determines data distribution (`category`)  
- **Clustering Columns:** Determines sort order (`product_id`)

## Chapter 3: Cassandra vs MongoDB
### 3.1 Architectural Comparison
```mermaid
graph TD
    A[Query Pattern] --> B[Cassandra: Write Optimized]
    A --> C[MongoDB: Flexible Queries]
    
    D[Consistency] --> E[Cassandra: Tunable]
    D --> F[MongoDB: Strong by Default]
    
    G[Data Model] --> H[Cassandra: Wide-Column]
    G --> I[MongoDB: Document]
```

### 3.2 Use Case Matrix
| Scenario | Cassandra | MongoDB |  
|----------|-----------|---------|

| Complex Transactions | ⚠️ Limited | ✅ Good |

## Chapter 4: DataStax Astra DB
### 4.1 What is Astra DB?
**Key Features:**  
- Fully managed Cassandra-as-a-Service  
- Serverless architecture  
- Multi-cloud support (AWS, GCP, Azure)  
- Built-in security & backups

**Architecture:**  
```mermaid
graph LR
    A[Client] --> B[Astra Gateway]
    B --> C[Coordinator Node]
    C --> D[Storage Layer]
    D --> E[Commit Log]
    D --> F[SSTables]
```

### 4.2 Astra DB Setup Guide
**Step 1: Create Database**  
1. Go to [Astra Portal](https://astra.datastax.com/)  
2. Choose "Create Database"  
3. Select cloud provider & region

**Step 2: Secure Connect Bundle**  
1. Navigate to "Connect" tab  
2. Download bundle  
3. Note credentials:
```
Client ID: 'token'
Client Secret: 'your-secret'
Secure Bundle Path: './secure-connect-dbname.zip'
```
